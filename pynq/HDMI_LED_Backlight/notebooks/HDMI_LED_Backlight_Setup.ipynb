# HDMI LED Backlight Setup

The HDMI LED Backlight overlay inserts IPs in the PYNQ video pipeline that calculate color data for WS2812C addressable LED strips on the back of the target display. The LED strips are driven by 4 pins of the PMODB board interface. The python interface of the overlay loads settings for the LED color calculation pipeline, while the overlay programmable logic does the heavy lifting of the color calculation and driving the WS2812C LEDs.

```python
import HDMI_LED_Backlight
```

## Setting up the overlay

Start the overlay, which downloads the programmable logic into the PL.

```python
overlay = HDMI_LED_Backlight.overlays.BacklightOverlay('HDMI_LED_Backlight.bit')
```

### Set up the HDMI video input

The default settings, which use a 24-bit BGR colorspace, are compatible with the LED pipeline. This will also pass the screen resolution to the LED driver.

```python
overlay.configure_hdmi()
```

### Set up the LED Pipeline

Set the number of LEDs in the strip on each side of the screen. The number of LEDs is a dictionary formatted as shown in the code.

```python
overlay.set_num_wc_leds({"top":120, "bottom":120, "left":100, "right":100})
```

Next, tell the overlay to calculate the remaining values it needs to run the pipeline. This is primarily defining regions at the edge of the screen assigned to each LED that are averaged to determine the LED's color.

```python
overlay.calc_bounds()
```

## Enable the overlay

Now the overlay is set up and ready to receive HDMI data. As long as the video input pipeline is running and the LED driver is enabled, then the LEDs will be driven.

```python
overlay.enable_led_output(True)
```

To disable the LED output, simply run the same function with `False` as the argument.

```python
overlay.enable_led_output(False)
```

## Enable HDMI Passthrough

Optionally, configure the HDMI output as a passthrough of the input

```python
hdmi_in = overlay.video.hdmi_in
hdmi_out = overlay.video.hdmi_out
hdmi_out.configure(hdmi_in.mode)
hdmi_in.start()
hdmi_out.start()
hdmi_in.tie(hdmi_out)
```
